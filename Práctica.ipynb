```python
import numpy as np

# Definimos la función objetivo
def funcion_objetivo(x):
    return -(x - 3)**2 + 10

# Algoritmo Hill Climbing
def hill_climbing():
    # Estado inicial
    x = 0  # Comenzamos en un valor arbitrario, por ejemplo 0
    paso = 0.1  # El tamaño del paso que damos
    iteraciones = 100  # Limite de iteraciones

    for _ in range(iteraciones):
        # Generamos dos vecinos (movimiento hacia la izquierda y la derecha)
        vecinos = [x - paso, x + paso]
        valores_vecinos = [funcion_objetivo(v) for v in vecinos]

        # Evaluamos qué vecino tiene el mejor valor de la función objetivo
        mejor_valor = max(valores_vecinos)
        if mejor_valor <= funcion_objetivo(x):
            # Si no encontramos un vecino mejor, terminamos
            print(f"Óptimo local encontrado en x = {x} con f(x) = {funcion_objetivo(x)}")
            break

        # Si encontramos un vecino mejor, nos movemos hacia él
        x = vecinos[valores_vecinos.index(mejor_valor)]

    return x

# Ejecutamos el algoritmo
resultado = hill_climbing()
print(f"Resultado final: x = {resultado}, f(x) = {funcion_objetivo(resultado)}")
```

```python
import numpy as np

# Matriz de distancias entre facultades (en minutos)
# Ejemplo de matriz de costos entre las 6 facultades
# Los valores son representativos de los tiempos de viaje entre las facultades
costos = np.array([
    [0, 10, 15, 20, 25, 30],  # Distancias desde la facultad 0
    [10, 0, 35, 25, 30, 20],  # Distancias desde la facultad 1
    [15, 35, 0, 30, 25, 10],  # Distancias desde la facultad 2
    [20, 25, 30, 0, 15, 5],   # Distancias desde la facultad 3
    [25, 30, 25, 15, 0, 10],  # Distancias desde la facultad 4
    [30, 20, 10, 5, 10, 0]    # Distancias desde la facultad 5
])

# Función para calcular el tiempo total de una ruta
def calcular_tiempo(ruta):
    tiempo_total = 0
    for i in range(len(ruta) - 1):
        tiempo_total += costos[ruta[i], ruta[i + 1]]  # Sumar el tiempo entre facultades consecutivas
    tiempo_total += costos[ruta[-1], ruta[0]]  # Volver al punto de inicio (facultad 0)
    return tiempo_total

# Función Hill Climbing para optimizar la ruta
def hill_climbing():
    ruta_actual = [0, 1, 2, 3, 4, 5, 0]  # Ruta inicial (comienza en la facultad 0)
    tiempo_actual = calcular_tiempo(ruta_actual)

    while True:
        vecinos = []  # Lista para almacenar las rutas vecinas

        # Generar vecinos intercambiando dos facultades consecutivas
        for i in range(1, len(ruta_actual) - 2):  # Evitamos el primer y último elemento
            vecino = ruta_actual[:]
            vecino[i], vecino[i + 1] = vecino[i + 1], vecino[i]  # Intercambiar dos facultades
            vecinos.append(vecino)

        mejor_ruta = ruta_actual
        mejor_tiempo = tiempo_actual

        # Evaluar los vecinos y elegir el mejor
        for vecino in vecinos:
            tiempo_vecino = calcular_tiempo(vecino)
            if tiempo_vecino < mejor_tiempo:
                mejor_ruta = vecino
                mejor_tiempo = tiempo_vecino

        # Si no encontramos ninguna mejora, terminamos
        if mejor_ruta == ruta_actual:
            break

        # Si encontramos una mejor ruta, la actualizamos
        ruta_actual = mejor_ruta
        tiempo_actual = mejor_tiempo

        print(f"Ruta mejorada: {ruta_actual} con tiempo: {tiempo_actual} minutos")

    return ruta_actual, tiempo_actual

# Ejecutar el algoritmo
ruta_optima, tiempo_optimo = hill_climbing()
print(f"\nRuta final óptima: {ruta_optima} con tiempo total: {tiempo_optimo} minutos")
```

```python
##APLICANDO HILL CLIMBING
##Función objetivo:
def funcion_objetivo(x):
    return -(x - 3)**2 + 10

##Variables de decisión:
# x
def hill_climbing(funcion, inicio, paso = 0.1, max_iter=1000):
    actual = inicio
    actual_valor = funcion_objetivo(actual)

    for _ in range(max_iter):
        ##evaluar vecinos inmediatos: izquierda y derecha
        vecino_izq = actual - paso
        vecino_der = actual + paso

        ##Encontrar el mejor vecino
        valor_izq = funcion_objetivo(vecino_izq)
        valor_der = funcion_objetivo(vecino_der)

        ##Encontrando el mejor vecino
        mejor_vecino = actual
        mejor_valor = actual_valor

        if valor_izq > mejor_valor:
            mejor_vecino = vecino_izq
            mejor_valor = valor_izq
        if valor_der > mejor_valor:
            mejor_vecino = vecino_der
            mejor_valor = valor_der

        #Si no hay más mejoras, hemos llegado al pico local
        if mejor_valor <= actual_valor:
            break
        ##Actualizar al mejor vecino
        actual = mejor_vecino
        actual_valor = mejor_valor
    return actual, actual_valor

#USÁNDOLO
inicio = 0 #punto inicial
resultado, valor = hill_climbing(funcion_objetivo, inicio)
print(f"Máximo aproximado en x = {resultado:.4f} con valor = {valor:.4f}")

```

```python
import math
import random

def funcion_objetivo(x):
    return -(x - 3)**2 + 10

def recocido_simulado(funcion, inicio, temp_inicial=10, temp_min=0.01, alfa=0.9, paso=0.1, max_iter=1000):
    actual = inicio
    actual_valor = funcion(actual)
    temperatura = temp_inicial

    iteracion = 0
    while temperatura > temp_min and iteracion < max_iter:
        # Generar vecino aleatorio (izquierda o derecha)
        if random.random() < 0.5:
            vecino = actual + paso
        else:
            vecino = actual - paso

        vecino_valor = funcion(vecino)
        cambio_calidad = vecino_valor - actual_valor

        # Si mejora, aceptar
        if cambio_calidad > 0:
            actual = vecino
            actual_valor = vecino_valor
        else:
            # Si empeora, aceptar con probabilidad P
            probabilidad = math.exp(cambio_calidad / temperatura)
            if random.random() < probabilidad:
                actual = vecino
                actual_valor = vecino_valor

        # Reducir temperatura
        temperatura *= alfa
        iteracion += 1

    return actual, actual_valor

# Uso del algoritmo
inicio = 0
resultado, valor = recocido_simulado(funcion_objetivo, inicio)

print(f"Máximo aproximado en x = {resultado:.4f} con valor = {valor:.4f}")
```
